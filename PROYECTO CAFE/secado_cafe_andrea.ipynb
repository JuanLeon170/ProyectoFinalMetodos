Cálculo de mermas y porcentaje de pérdida de peso durante el proceso de secado.

```python
# Variables del proceso de secado de café
peso_inicial_kg = 100.0  # Peso húmedo
peso_final_kg = 82.5     # Peso seco después del proceso

# Operaciones básicas
merma_kg = peso_inicial_kg - peso_final_kg
porcentaje_merma = (merma_kg / peso_inicial_kg) * 100

# Resultados
print("--- CONTROL DE SECADO DE CAFÉ ---")
print(f"Peso inicial: {peso_inicial_kg} kg")
print(f"Peso final: {peso_final_kg} kg")
print(f"Perdida por merma: {merma_kg} kg")
print(f"Porcentaje de merma: {porcentaje_merma:.1f}%")
```
